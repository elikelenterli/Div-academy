### Map , Reduce ,Filter, Lambda

```python
numbers = [2, 4, 6, 8]

result = map (lambda x: x ** 2, numbers)
print(list(result))
```

```python
numbers = [2, 7, 4, 9, 3, 8, 1, 6]
result = filter(lambda x : x > 5 ,numbers)

print (list(result))
```

```python
from functools import reduce

numbers = [2, 3, 4, 5]

result = reduce(lambda x,y: x * y , numbers)
print (result)

```

```python
numbers = [12, 5, 18, 7, 24, 9, 30, 11, 6]

result = map(lambda x: x ** 2, filter(lambda x: x % 3 == 0 ,numbers ))

print(list(result))
```

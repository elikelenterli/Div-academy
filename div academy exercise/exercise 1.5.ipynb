### Exercise 3

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def describe(self):
        return f"Bu {self.year} {self.make} {self.model}-dir"
    
    
my_car = Car(2020,"Ford","Mustang")
print(my_car.describe())


class ElectricCar(Car):
    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year)
        self.battery_size = battery_size


my_car = ElectricCar("Tesla", "Model S", 2022, 100)

print(my_car.make)
print(my_car.model)
print(my_car.year)
print(my_car.battery_size)
```

### Exercise 13

```python
class Car:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    @classmethod
    def from_string(cls, car_string):
        make, model = car_string.split("-")
        return cls(make, model)


car1 = Car.from_string("Toyota-Corolla")

print(car1.make)
print(car1.model)
```

### Exercise 14

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return self.first_name + " " + self.last_name

    @full_name.setter
    def full_name(self, name):
        self.first_name, self.last_name = name.split(" ")


person = Person("eli", "kelenerli")

print(person.full_name)

person.full_name = "Murad eliyev"

print(person.first_name)
print(person.last_name)
print(person.full_name)
```

```python
def elifba_yarat():
	import string

	orta = len(string.ascii_letters) // 2
	kicik_herfler = string.ascii_letters[:orta]
	boyuk_herfler = string.ascii_letters[orta:]

	bosh_dict={}

	for ind in range(0, orta):
		bosh_dict[kicik_herfler[ind]] = boyuk_herfler[ind]
	return bosh_dict

elifba = elifba_yarat()
```

```python
def ilk_herfi_boyut(soz):
	ilk_herf = soz[0] # 'a'
	elifba = elifba_yarat()
	if ilk_herf in elifba:
		ilk_herf = elifba[ilk_herf] 	# ilk_herf = elifba['a'] = 'A'

	return ilk_herf + soz[1:]

ilk_herfi_boyut('hesen')
```
